### 9. Test Your Fine-tuned Model (Inference Demo)
Use this cell to interact with your fine-tuned model. You can enter routing scenarios or questions and see how your newly trained model handles them.

```python
from unsloth import FastLanguageModel
import torch

# Configure model for inference
FastLanguageModel.for_inference(model)

# Clear tool calling templates and configurations that intercept JSON
tokenizer.chat_template = None
if hasattr(model, "generation_config"):
    model.generation_config.forced_bos_token_id = None
    model.generation_config.forced_eos_token_id = None
    if hasattr(model.generation_config, "tools"):
        model.generation_config.tools = None

def ask_model(user_message, system_instruction="You are a routing assistant for Tata Steel's internal data-query system. Decide which agent should handle the question."):
    # Define the precise ChatML format the adapter was trained on
    prompt = f"<|im_start|>system\n{system_instruction}<|im_end|>\n<|im_start|>user\n{user_message}<|im_end|>\n<|im_start|>assistant\n"
    inputs = tokenizer([prompt], return_tensors="pt").to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            input_ids=inputs.input_ids,
            attention_mask=inputs.attention_mask,
            max_new_tokens=150,
            temperature=0.1,
            top_p=0.9,
            use_cache=True,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.encode("<|im_end|>")[0]
        )

    prompt_length = inputs.input_ids.shape[1]
    decoded_output = tokenizer.decode(outputs[0][prompt_length:], skip_special_tokens=True)

    # Manual cleanup of remaining chat wrappers
    for delimiter in ["<|im_start|>", "<|im_end|>", "assistant", "system", "user"]:
        decoded_output = decoded_output.replace(delimiter, "")

    return decoded_output.strip()

# --- Interactive Testing Interface ---
print("Model inference interface initialized successfully!")
test_out = ask_model("What is the total quantity and value of 'Flats' classification inventory today?")
print(f"\nDry Run Test Output:\n{test_out}\n" + "="*40)
```

### 10. Quantitative Evaluation (Loss, Perplexity, and Ground-Truth Comparison)
This cell evaluates the fine-tuned model on the **held-out test set** to calculate perplexity and loss. It then prints a side-by-side comparison of the model's responses against the target ground truth.

```python
import torch
import numpy as np
from tqdm import tqdm

# Ensure model is in evaluation mode
model.eval()

total_loss = 0.0
total_tokens = 0

print("Calculating loss and perplexity using standard evaluation loop...")
loss_fn = torch.nn.CrossEntropyLoss(reduction="sum")

with torch.no_grad():
    for row in test_rows:
        text = tokenizer.apply_chat_template(row["messages"], tokenize=False)
        inputs = tokenizer(text, return_tensors="pt").to("cuda")

        # Retrieve raw logits to compute loss directly without compilation flags
        outputs = model(input_ids=inputs["input_ids"])
        logits = outputs.logits

        # Shift so next-token prediction is evaluated
        shift_logits = logits[..., :-1, :].contiguous()
        shift_labels = inputs["input_ids"][..., 1:].contiguous()

        # Flatten the tokens
        loss = loss_fn(shift_logits.view(-1, shift_logits.size(-1)), shift_labels.view(-1))

        num_tokens = shift_labels.numel()
        total_loss += loss.item()
        total_tokens += num_tokens

average_loss = total_loss / total_tokens
perplexity = np.exp(average_loss)

print("\n================ EVALUATION METRICS ================")
print(f"Test Cross-Entropy Loss : {average_loss:.4f}")
print(f"Test Perplexity         : {perplexity:.4f}")
print("====================================================\n")

# Side-by-side verification
print("\n--- Qualitative Side-by-Side Comparison ---")
for i, row in enumerate(test_rows[:3]):
    system_prompt = ""
    user_prompt = ""
    ground_truth = ""

    for msg in row["messages"]:
        if msg["role"] == "system":
            system_prompt = msg["content"]
        elif msg["role"] == "user":
            user_prompt = msg["content"]
        elif msg["role"] == "assistant":
            ground_truth = msg["content"]

    predicted = ask_model(user_prompt, system_instruction=system_prompt)

    print(f"\n[Test Case #{i+1}]")
    print(f"User Query  : {user_prompt}")
    print(f"Expected    : {ground_truth}")
    print(f"Model Output: {predicted}")
    print("="*60)
```
